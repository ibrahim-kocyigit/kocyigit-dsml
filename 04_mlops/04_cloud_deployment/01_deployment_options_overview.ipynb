## 1. Deployment Categories in Detail

### 1.1. PaaS (Platform as a Service)

- **You give:** Your source code or Docker image.
- **They handle:** Servers, networking, SSL, scaling, OS updates
- **When to use:** Quick deploys, client prototypes, small-to-medium APIs, projects where you don't want to manage infrastructure.

| **Platform** | **Free Tier** | **Docker Support** | **Deploy From Github** | **Notes** |
| :--- | :--- | :---: | :---: | :--- |
| **[Railway](https://railway.com/)** | $5/mo credit | ✅ | ✅ | Excellent DX, simple pricing, great for small APIs |
| **[Render](https://render.com/)** | Free with sleep | ✅ | ✅ | Free tier sleeps after inactivity (cold starts) |
| **[Fly.io](https://fly.io/)** | Free tier available | ✅ | ✅ | Global edge deployment, more DevOps-oriented |
| **[Heroku](https://www.heroku.com/)** | No free tier (paid only) | ✅ | ✅ | The original PaaS, mature but expensive |


### 1.2. CaaS (Container as a Service)

- **You give:** A Docker image.
- **They handle:** Running the container, scaling, networking.
- **When to use:** Docker-based apps that need auto-scaling. Ideal when traffic is unpredictable (scale-to-zero saves money).

| **Platform** | **Free-Tier** | **Pay Model** | **Notes** |
| :--- | :--- | :--- | :--- |
| **Google Cloud Run** | Generaous free tier | Per-request | Scales to zero (no cost when idle). Excellent for bursty traffic. |
| **AWS App Runnder** | No free tier | Per-instance-hour | Simple but less cost-effective for low traffic. |
| **Azure Container Apps** | Free tier available | Per-request | Good if client already uses Azure | 


### 1.3. IaaS (Infrastructure as a Service)
- **You give:** Everything. You manage the OS, runtime, networking, security.
- **The give:** A virtual machine.
- **When to use:** You need full control (custom runtimes, GPU access, specific OS configurations). Overkill for most ML API deployments.

| Platform | Entry Cost | Notes |
| :--- | :--- | :--- |
| **AWS EC2** | ~$5/mo (t3.micro) | Full control, steep learning curve |
| **Google Compute Engine** | ~$5/mo | Same idea, Google ecosystem |
| **DigitalOcean Droplets** | $4/mo | Simpler UX, good docs, popular with freelancers |

### 1.4. Serverless Functions

- **You give:** A single function.
- **They handle:** Everything.
- **When to use:** Lightweight tasks, event-driven processing. **Not ideal for ML** because of cold start times (loading a model from scratch on every cold start adds seconds of latency).

| Platform | Free Tier | Notes |
| :--- | :--- | :--- |
| **AWS Lambda** | 1M requests/mo free | Cold starts can be slow for ML models |
| **Google Cloud Functions** | 2M invocations/mo free | Same cold start issue |

## 2. Key Decision Factors

When choosing a platform for a client project, consider these factors:

| **Factor** | **Question to Ask** | **Impact** |
| :--- | :--- | :--- |
| **Traffic Volume** | How many requests per day/hour? | Low traffic → PaaS/scale-to-zero CaaS. High traffic → CaaS with auto-scaling. |
| **Latency Requirements** | Does the client need sub-100ms responses? | Serverless is out (cold starts). CaaS or always-on PaaS. |
| **Budget** | How much is the client willing to spend? | Free tier → Railway/Render/Cloud Run. Production budget → CaaS. |
| **Model Size** | How large is the model file? | Small (<100MB) → any platform. Large (>1GB) → IaaS or CaaS with more memory. | 
| **Client Existing Stack** | Does the client already use AWS/GCP/Azure? | Use what they already have to avoid multi-cloud complexity. |
| **Your Expertise** | How much DevOps do your want to manage? | Minimal → PaaS. Comfortable → CaaS. Expert → IaaS |
| **GPU Needs** | Does the model need GPU for inference? | GPU → IaaS (EC2 GPU instances) or specialized platforms (Replicate, Modal). |

## 3. Decision Flowchart


Here's a practical flowchart for choosing a deployment strategy:

```
Does the model need a GPU?
├── Yes → IaaS with GPU (EC2 p3/g4 instances) or Modal/Replicate
└── No
    ├── Is the client already on AWS/GCP/Azure?
    │   ├── Yes → Use their CaaS (Cloud Run / App Runner / Container Apps)
    │   └── No
    │       ├── Is traffic unpredictable or low?
    │       │   ├── Yes → Google Cloud Run (scale-to-zero) or Railway
    │       │   └── No → Railway / Render / Fly.io
    │       └── Is it a quick prototype / demo?
    │           ├── Yes → Railway or Render (fastest to deploy)
    │           └── No → Evaluate CaaS options based on budget
```

**For this learning path**, we'll use **Railway** because:
- Zero cloud provider setup required (no AWS/GCP account needed)
- Deploys directly from a GitHub repository
- Auto-detects Dockerfiles
- Has a free tier ($5/month credit)
- Takes less than 5 minutes to deploy

## 4. Cost Awareness

As a freelancer, you need to understand cloud costs so you can advise clients (and not get surprised by your own bills).

### Typical Monthly Costs for a Small ML API

| Platform | ~Cost for Low Traffic (<1000 req/day) | Notes |
| :--- | :--- | :--- |
| Railway | $0–5 | Free credit covers most small projects |
| Render (free tier) | $0 | Service sleeps after 15 min of inactivity |
| Google Cloud Run | $0–3 | Scale-to-zero, pay per request |
| Fly.io | $0–5 | Free tier includes small VMs |
| AWS EC2 (t3.micro) | ~$8 | Always on, even with no traffic |
| Heroku (Basic) | $7 | No free tier anymore |

### Cost Traps to Watch For

- **Always-on instances:** Platforms like EC2 charge even when no one is using the API.
- **Egress fees:** Some cloud providers charge for outbound data transfer (sending responses).
- **Model size = memory = cost:** Large models need more RAM, which means higher-tier instances.
- **Logging and monitoring:** Some platforms charge for log storage and metrics.

## 5. Summary

| Category | Best For | Complexity | Cost |
| :--- | :--- | :---: | :---: |
| **PaaS** | Quick deploys, prototypes, small APIs | Low | Low |
| **CaaS** | Docker apps, auto-scaling, production | Medium | Low–Medium |
| **IaaS** | Full control, GPU, complex setups | High | Variable |
| **Serverless** | Event-driven, lightweight tasks | Low | Very Low |

**The freelancer's rule:** Start with PaaS. Move to CaaS when the project demands scaling. Only touch IaaS when you need GPU or full server control.
