# Regression Modeling in Python

This notebook applies the concepts from the Week 2 lectures. We will perform a case study using the NHANES dataset to build, interpret, and visualize both linear and logistic regression models using `statsmodels` in Python.

### Setup and Data Preparation
First, we import the necessary libraries and load the NHANES dataset. We will then clean and preprocess the data, preparing it for our regression analyses.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import statsmodels.api as sm
import numpy as np

# Read the data file 
df = pd.read_csv("./data/nhanes_2015_2016.csv")

# Keep only the columns of interest
vars = ["BPXSY1", "RIDAGEYR", "RIAGENDR", "RIDRETH1", "DMDEDUC2", "BMXBMI", "SMQ020"]
df = df[vars].dropna()

# --- Data Recoding and Cleaning ---

# Recode gender to a more readable string variable
df["RIAGENDRx"] = df.RIAGENDR.replace({1: "Male", 2: "Female"})

# Recode education level to a string variable, treating '7' and '9' as missing
df["DMDEDUC2x"] = df.DMDEDUC2.replace({1: "<9", 2: "9-11", 3: "HS/GED", 4: "Some college/AA", 5: "College",
                                       7: np.nan, 9: np.nan})

# Recode smoking status into a binary variable (1=smoker, 0=non-smoker)
df["smq"] = df.SMQ020.replace({1: 1, 2: 0, 7: np.nan, 9: np.nan})

# Display the first few rows of the cleaned data
df.head()
```


---

## 1. Linear Regression Modeling in Python
Our goal is to model systolic blood pressure (`BPXSY1`) as a function of age, gender, and BMI.

### Fitting a Multiple Linear Regression Model
We use `statsmodels.OLS.from_formula()` to fit our model. This formula specifies that `BPXSY1` is the dependent variable, and `RIDAGEYR`, `RIAGENDRx`, and `BMXBMI` are the independent variables.

```python
# Fit the model
model = sm.OLS.from_formula("BPXSY1 ~ RIDAGEYR + RIAGENDRx + BMXBMI", data=df)
result = model.fit()

# Print the model summary
print(result.summary())
print("="*78)
```
